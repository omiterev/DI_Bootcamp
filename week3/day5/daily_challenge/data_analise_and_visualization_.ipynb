```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

superstore=pd.read_excel('US Superstore data.xls')
superstore.info()
superstore.describe()
superstore.isnull().sum()
```

```python
superstore.duplicated().sum()
```

```python
superstore.drop(columns=['Row ID','Country'],inplace=True)
```

```python
import datetime
superstore['Order Date']=pd.to_datetime(superstore['Order Date'])
superstore['Ship Date']=pd.to_datetime(superstore['Ship Date'])
```

```python
state_sales=superstore.groupby('State',as_index=False)['Sales'].sum().sort_values('Sales',ascending=False)
state_sales.reset_index(drop=True,inplace=True)
print(f"Top state sales {state_sales.loc[0,'State']} - ${state_sales.loc[0,'Sales']:,.1f}")
```

```python
state_sales_profit=superstore.groupby('State',as_index=False)[['Sales','Profit']].sum().sort_values(['Sales','Profit'],ascending=[False, False])
state_sales_profit.set_index('State',inplace=True)
# state_sales_profit.head()
ny_sales, ny_profit=state_sales_profit.loc['New York',['Sales','Profit']].round(1)
ca_sales, ca_profit=state_sales_profit.loc['California',['Sales','Profit']].round(1)
print (f'New York sales {ny_sales:,.1f}, profit {ny_profit:,.1f}')
print (f'California sales {ca_sales:,.1f}, profit {ca_profit:,.1f}')
print (f'California made {(ca_sales-ny_sales):,.1f} more sales than New York')
print (f'California made {(ca_profit-ny_profit):,.1f} more profit than New York')
```

```python
customer=superstore.loc[superstore['State']=='New York']
customer=customer.groupby(['Customer ID','Customer Name'],as_index=False)['Sales'].sum()
customer.sort_values('Sales',ascending=False,inplace=True)
customer.reset_index(drop=True,inplace=True)
print(f"Top customer in New York is {customer.loc[0,'Customer Name']} - ${customer.loc[0,'Sales']:,.1f}")
```

```python
# state_perf = (superstore.groupby('State', as_index=False)
#              .agg(Sales=('Sales','sum'),
#                   Profit=('Profit','sum')))

profitability=superstore.groupby("State", as_index=False)[['Sales','Profit']].sum()
profitability.reset_index(drop=True,inplace=True)
profitability['Profitability']=(profitability['Profit']/profitability['Sales'])*100
profitability.sort_values(['Profitability'],ascending=False, inplace=True)
profitability.reset_index(drop=True,inplace=True)
print(profitability[['State','Profitability']].head(5).to_string(index=False, formatters={'Profitability': '{:.1f}%'.format}))
print(profitability[['State','Profitability']].tail(5).to_string(index=False, formatters={'Profitability': '{:.1f}%'.format}))
profitability_negative=profitability[profitability['Profitability']<0]
profitability_negative
print(f'States with negative total profit:', ",  ".join(profitability_negative["State"]))
# profitability.tail(20)
```

```python
# Top 20 customers
customer_profit=superstore.groupby(['Customer ID','Customer Name'],as_index=False)[['Sales','Profit',]].sum()
customer_p=customer_profit.sort_values(['Profit'],ascending=False).reset_index(drop=True)
customer_s=customer_profit.sort_values(['Sales'],ascending=False).reset_index(drop=True)
sales=superstore['Sales'].sum()
profit=superstore['Profit'].sum()
print(f'Total sales {sales:,.1f}')
print(f'Total profit {profit:,.1f}')
customer_p['Profit%']=(customer_p['Profit']/profit)*100
customer_s['Sales%']=(customer_s['Sales']/sales)*100
customer20=int((len(customer_p)/100)*20)
top20=customer_p.head(customer20)
top20_s=customer_s.head(customer20)
customer_p_sum_per=top20['Profit%'].sum()
customer_p_sum=top20['Profit'].sum()
customer_s_sum=top20_s['Sales'].sum()
customer_s_sum_per=top20_s['Sales%'].sum()

# print(customer_p.head(int(customer20)))
print(f'Top 20% customers bring {customer_p_sum_per:,.1f}% of profit, {customer_p_sum:,.1f}$')
print(f'Top 20% customers bring {customer_s_sum_per:,.1f}% of sales, {customer_s_sum:,.1f}$')

print(f'Top 20 customers by Sales\n{customer_s.head(20)}')
print(f'Top 20 customers by Profit\n{customer_p.head(20)}')
```

```python
#Top 20 Sub-Category
import math
sub_category=superstore.groupby('Sub-Category',as_index=False)[['Sales','Profit']].sum().sort_values('Profit',ascending=False).reset_index(drop=True)
sub_category
top20index=math.ceil((len(sub_category)/100)*20)
top20_category=sub_category.head(top20index)
top20_category['Profit%']=(top20_category['Profit']/profit)*100
top20_profit_sum=top20_category['Profit%'].sum()
print(f'Top 20% categories bring {top20_profit_sum:,.1f}% of profit')
print(top20_category.head())
```

```python
#Citys Sales Profit and Profitability
df_city=superstore.groupby('City',as_index=False)[['Sales','Profit']].sum()
city_sales=df_city.sort_values('Sales',ascending=False).reset_index(drop=True)
city_profit=df_city.sort_values('Profit',ascending=False).reset_index(drop=True)
city_profit['Profit%']=(city_profit['Profit']/profit)*100
city_sales['Sales%']=(city_sales['Sales']/sales)*100
city_profit['Profitability']=(city_profit['Profit']/city_profit['Sales'])*100
city_sales['Profitability']=(city_sales['Profit']/city_sales['Sales'])*100
top20_sales=city_sales.head(20)
top20_profit=city_profit.head(20)
print(f' Top 20 cities by sale\n{city_sales.head(20)}')
print(f' Top 20 cities by profit\n{city_profit.head(20)}')

```

```python
sales_only=(top20_sales[top20_sales['City'].isin(top20_profit['City'])])
sales_only_ex=(top20_sales[~top20_sales['City'].isin(top20_profit['City'])])

print(f'From top 20 cities sales only {len(sales_only)} ({", ".join(sales_only["City"])}) is in top 20 cities by profit')
print(f'In {len(sales_only_ex)} cities ({", ".join(sales_only_ex["City"])}) sales are high but low profit')
```

```python
city_perf = (superstore.groupby('City', as_index=False)
             .agg(Sales=('Sales','sum'), Profit=('Profit','sum')))
city_perf['Profit%'] = np.where(
    city_perf['Sales'] > 0,
    city_perf['Profit'] / city_perf['Sales'] * 100,
    np.nan
)

# 2) Топ-20 по продажам и по прибыли
top20_sales  = city_perf.sort_values('Sales',  ascending=False).head(20)
top20_profit = city_perf.sort_values('Profit', ascending=False).head(20)

# 3) Анализ profitability среди городов из обоих топов
union_cities = pd.Index(top20_sales['City']).union(top20_profit['City'])
union_perf = (city_perf.set_index('City')
                        .loc[union_cities]
                        .reset_index())

# Размах маржи и убытки
pm = union_perf['Profit%']
spread = pm.max() - pm.min()
loss = union_perf[union_perf['Profit'] < 0][['City','Sales','Profit','Profit%']]

print(f"Cities considered: {len(union_perf)}")
print(f"Profitability spread (max - min): {spread:.2f} percentage points")

if len(loss):
    print("\nLoss-making cities within the top lists:")
    print(loss.sort_values('Profit')
             .to_string(index=False,
                        formatters={'Sales':'{:,.0f}'.format,
                                    'Profit':'{:,.0f}'.format,
                                    'Profit%':'{:.2f}%'.format}))
else:
    print("\nNo loss-making cities within the top lists.")

# Топ/низ по марже в объединённом списке
print("\nTop 5 cities by Profit Margin (within union):")
print(union_perf.sort_values('Profit%', ascending=False)
      .head(5)[['City','Sales','Profit','Profit%']]
      .to_string(index=False,
                 formatters={'Sales':'{:,.0f}'.format,
                             'Profit':'{:,.0f}'.format,
                             'Profit%':'{:.2f}%'.format}))

print("\nBottom 5 cities by Profit Margin (within union):")
print(union_perf.sort_values('Profit%', ascending=True)
      .head(5)[['City','Sales','Profit','Profit%']]
      .to_string(index=False,
                 formatters={'Sales':'{:,.0f}'.format,
                             'Profit':'{:,.0f}'.format,
                             'Profit%':'{:.2f}%'.format}))

# 4) Несовпадения списков: «много продаж, но не топ по прибыли» и наоборот
sales_only = (top20_sales[~top20_sales['City'].isin(top20_profit['City'])])
profit_only = (top20_profit[~top20_profit['City'].isin(top20_sales['City'])])

print(f"\nTop-20 Sales only (count={len(sales_only)}):")
print(sales_only[['City','Sales','Profit%']]
      .to_string(index=False,
                 formatters={'Sales':'{:,.0f}'.format,'Profit%':'{:.2f}%'.format}))

print(f"\nTop-20 Profit only (count={len(profit_only)}):")
print(profit_only[['City','Profit','Profit%']]
      .to_string(index=False,
                 formatters={'Profit':'{:,.0f}'.format,'Profit%':'{:.2f}%'.format}))
```

What you can see on the charts

Geography (Plotly choropleth – just noting):

Sales are highly concentrated in CA (clear outlier) and strong on the East Coast (e.g., NY, MA).

Central/Mountain states are weaker → obvious targets for growth or different go-to-market.

Sub-Category sales (bar chart):

Phones and Chairs lead by a wide margin; Appliances/Bookcases trail.

Portfolio looks long-tailed → a few SKUs drive most revenue.

Profit vs. Discount (Seaborn scatter with bubble size = Sales):

Clear negative relationship: deeper discounts → profit drops; beyond ~40–50% discounts profit is mostly negative.

Technology has the biggest bubbles (large sales) with wide profit variance → discount policy there is especially sensitive.

Visible clusters by Category help spot where discounts hurt the most.

Sales by Year (Matplotlib line):

Slight dip from 2014→2015, then strong YoY growth through 2017.

Momentum suggests demand + assortment/pricing worked better post-2015.

Matplotlib vs Seaborn

Seaborn: faster and nicer “out of the box”. With one function you get color by category, point sizes, a normal legend. Ideal for quick insights.

Matplotlib: more fuss, but full control over signatures, formats, lines, style. Good for “finishing up to a presentation”.

Plotly (as a map): when you need interactivity and hover hints.
