```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
temperature_data=np.random.uniform(low=-5,high=35,size=(10,12)).round(1)
temperature=pd.DataFrame(temperature_data,index=['City1','City2','City3','City4','City5','City6','City7','City8','City9','City10'],columns=['Jan','Feb','Mar','Apr','May','Jun','Jul','Aug','Sep','Oct','Nov','Dec'])
temperature
```

```python
temperature_year=np.mean(temperature, axis=1)
temperature_year
```

```python
print(f'City with the highest average temperature is {temperature_year.idxmax()}')
print(f'City with the lowest average temperature is {temperature_year.idxmin()}')
```

```python
months = temperature.columns
x = np.arange(len(months))
```

```python
plt.figure(figsize=(15, 6), dpi=120)
colors = plt.cm.GnBu(np.linspace(0.25, 0.95, len(temperature.index)))
for t,city in enumerate(temperature.index):
  y=temperature.loc[city].values
  ls='--' if city in {'City3','City8','City5'} else '-'
  plt.plot(x,y,label=city,linestyle=ls, color=colors[t])
plt.xticks(x, months)
plt.ylabel('Temperature (°C)')
plt.xlabel('Month')
plt.title('Monthly Temperature Trends for Each City')
plt.legend(loc="upper left", bbox_to_anchor=(1, 1))
plt.grid(True)
plt.tight_layout()
plt.show()
```
